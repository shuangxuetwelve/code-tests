### Imports and Environment Variables

````python
import os
import requests
from pprint import pprint as pp

from dotenv import load_dotenv
load_dotenv()

BRIGHT_DATA_API_KEY = os.getenv("BRIGHT_DATA_API_KEY")
if BRIGHT_DATA_API_KEY == None:
  raise Exception("Bright Data API key is not set")
print(f"Bright Data API key: *******************************")
````

### Get YouTube Profile by URL

````python
# 必须是正确的channel URL，不能是https://www.youtube.com/@bling1761/videos这样的URL
target_url = "https://www.youtube.com/@bling1761"

url = "https://api.brightdata.com/datasets/v3/scrape?dataset_id=gd_lk538t2k2p1k3oos71&include_errors=true"

headers = {
    "Authorization": f"Bearer {BRIGHT_DATA_API_KEY}",
    "Content-Type": "application/json"
}

data = [
  {
    "url": target_url,
  }
]

response = requests.post(url, headers=headers, json=data)
pp(response.json())
````

### Get YouTube Profiles by Keywords

````python
keywords = "鱼闪闪"

url = "https://api.brightdata.com/datasets/v3/scrape?dataset_id=gd_lk538t2k2p1k3oos71&include_errors=true&type=discover_new&discover_by=keyword"

headers = {
    "Authorization": f"Bearer {BRIGHT_DATA_API_KEY}",
    "Content-Type": "application/json"
}

data = [
  {
    "keyword": keywords,
  }
]

response = requests.post(url, headers=headers, json=data)
pp(response.json())
````
