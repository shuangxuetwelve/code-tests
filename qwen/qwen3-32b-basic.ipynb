```python
import os
from pprint import pprint as pp

from dotenv import load_dotenv
load_dotenv()

from groq import Groq

GROQ_API_KEY = os.getenv("GROQ_API_KEY")

client = Groq(api_key=GROQ_API_KEY)
```

```python
messages = [
  {
    "role": "system",
    "content": "You are a helpful assistant."
  },
  {
    "role": "user",
    "content": "What is the capital of France?"
  }
]

response = client.chat.completions.create(
  model="qwen/qwen3-32b",
  messages=messages,
  reasoning_effort="default", # Options are "default" and "none". Defaults to "default"
  reasoning_format="raw", # Options are "raw", "parsed", "hidden". Defaults to "raw".
)
```

```python
pp(response.choices[0].message.content)
print("usage", response.usage)
```

```python
messages.extend([
  {
    "role": "assistant",
    "content": response.choices[0].message.content, # 即使content中包含<thinking> block，也会被忽略，不算是input tokens。
  },
  {
    "role": "user",
    "content": "What is the most famous city in France?",
  }
])

response = client.chat.completions.create(
  model="qwen/qwen3-32b",
  messages=messages,
)
```

```python
pp(response.choices[0].message.content)
print("usage", response.usage)
```
